# LTspice waveform EDA

Run `python simulate.py` and `python extract_features.py` from the repository root before opening this notebook. All figures below use generated LTspice results; there is no bundled fabricated dataset.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

manifest_path = Path('../data/ltspice/manifest.csv')
features = pd.read_csv('../data/processed/features.csv')
manifest = pd.read_csv(manifest_path)
features.head()
```

## Representative output waveforms by fault class

```python
fig, ax = plt.subplots(figsize=(12, 5))
for label, group in manifest.groupby('fault_label'):
    trace = pd.read_csv(manifest_path.parent / group.iloc[0]['waveform_path'])
    ax.plot(trace['time_s'], trace['output_v'], label=label, linewidth=1)
ax.set(xlabel='Time (s)', ylabel='Output voltage (V)', title='LTspice transient waveforms')
ax.grid(alpha=.25); ax.legend(ncol=2); plt.show()
```

## Feature distributions and correlation

```python
feature_cols = ['peak_amplitude_v', 'rms_v', 'rise_time_s', 'settling_time_s',
                'thd_percent', 'zero_crossings', 'std_dev_v', 'skewness']
fig, axes = plt.subplots(4, 2, figsize=(14, 14))
for ax, name in zip(axes.flat, feature_cols):
    sns.boxplot(data=features, x='fault_label', y=name, ax=ax)
    ax.tick_params(axis='x', rotation=35)
fig.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(9, 7))
sns.heatmap(features[feature_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Feature correlation'); plt.tight_layout(); plt.show()
```

Do not interpret these plots as hardware validation. The current circuit uses a bounded behavioral op-amp model and randomized passive-component tolerances.
